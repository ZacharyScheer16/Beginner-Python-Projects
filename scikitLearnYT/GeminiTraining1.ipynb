```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

names = ['mouse', 'elephant', 'giraffe']
X = [
    [1, 1], [2, 1],      # Mice
    [10, 12], [11, 15],  # Elephants
    [19, 4], [20, 5]     # Giraffes
]
y = [0, 0, 1, 1, 2, 2]

# 1. Standard Scaling
scalar = StandardScaler()
X_scaled = scalar.fit_transform(X)

# --- THE TRICK: MANUAL WEIGHTING ---
# We multiply the HEIGHT column (index 0) by 10
# This makes Height 10x more "important" than Weight
X_weighted = X_scaled.copy()
X_weighted[:, 0] = X_weighted[:, 0] * 10
# ----------------------------------

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_weighted, y)

# 2. Test Animals
test_animal = [[5, 5], [12, 12], [15, 5], [1, 20]]
test_scaled = scalar.transform(test_animal)

# --- MUST APPLY THE SAME TRICK TO TEST DATA ---
test_weighted = test_scaled.copy()
test_weighted[:, 0] = test_weighted[:, 0] * 10
# ----------------------------------------------

guess = model.predict(test_weighted)

for i in range(len(test_animal)):
    print(f"Animal {test_animal[i]} is a {names[guess[i]]}")
```
